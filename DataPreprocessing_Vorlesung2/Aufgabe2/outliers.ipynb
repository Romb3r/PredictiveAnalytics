First, look at the dataset of adult body weights. 

```python
#import packages
import numpy as np
import matplotlib.pyplot as plt
import random

# define your dataset of normal distribution adults
normsample_adults = np.random.normal(80, 20, 30)
outliers = np.array([170, 10, 25, 135])
data = np.append(normsample_adults, outliers)
print(data)
```

Second, visualize the data as boxplot by using the **Seaborn** data visualization library  ([Link to Seaborn Boxplot Doku](https://seaborn.pydata.org/generated/seaborn.boxplot.html)).

```python
# Visualize the data as boxplot
import seaborn as sns
sns.boxplot(x=data)

```

Third, print *Q1*, *Median*, *Q3*, and *IQR* and compare the values with your boxplot. There are differnt ways to do this.

```python
# print Q1
q1_value = np.percentile(data, 25)
print(q1_value)

# print Median
median_value = np.percentile(data,50)
print(median_value)

# print q3

q3_value = np.percentile(data, 75)
print(q3_value)

# Print IQR distance between (Q3 and Q1)
iqr_value = q3_value - q1_value
print(f"IQR: {iqr_value}")
```

Write a function `find_outliers(data)` and `find_extreme_outliers`(data) to detect outlier and extreme outlier in your dataset.

```python

def find_outliers(data_frame:list[int]):
    normal_outliers=[] #type: list[int]
    extreme_outliers=[]  #type: list[int]
    for i in data_frame:
        if i<q1_value-(3*iqr_value) or i>q3_value+(3*iqr_value):
            extreme_outliers.append(float(i))
        elif i<q1_value-(1.5*iqr_value) or i >q3_value+(1.5*iqr_value):
            normal_outliers.append(float(i))
    
    return f"Normal Outliers: {normal_outliers}\nExtreme Outliers:{extreme_outliers}"
    
```
